```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from ucimlrepo import fetch_ucirepo

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, classification_report, ConfusionMatrixDisplay,
    roc_auc_score
)

sns.set_theme(style="whitegrid")

# Download the real UCI Online Retail dataset
dataset = fetch_ucirepo(id=352)
df = dataset.data.features.copy()

df.columns = [str(c).strip() for c in df.columns]

print("Dataset downloaded successfully!")
print("Dataset shape:", df.shape)
print("Columns:", df.columns.tolist())

display(df.head())
```

```python
# Reconstruct using dataset.data.original to ensure 'InvoiceNo' is present
import pandas as pd

df = dataset.data.original.copy()
df.columns = [str(c).strip() for c in df.columns]

# Convert columns to the correct data types
if "InvoiceNo" in df.columns:
    df["InvoiceNo"] = df["InvoiceNo"].astype(str).str.strip()
else:
    df["InvoiceNo"] = "Unknown"

df["InvoiceDate"] = pd.to_datetime(
    df["InvoiceDate"], errors="coerce"
)
df["Quantity"] = pd.to_numeric(
    df["Quantity"], errors="coerce"
)
df["UnitPrice"] = pd.to_numeric(
    df["UnitPrice"], errors="coerce"
)

# Remove records missing essential information
df = df.dropna(
    subset=["InvoiceNo", "InvoiceDate", "Quantity", "UnitPrice"]
)

# Remove blank invoice numbers
df = df[df["InvoiceNo"] != ""].copy()

# Create a return/cancellation label
df["returned"] = (
    df["InvoiceNo"].str.upper().str.startswith("C")
    | (df["Quantity"] < 0)
).astype(int)

# Calculate transaction value
df["total_value"] = (
    df["Quantity"].abs() * df["UnitPrice"].abs()
)

print("Cleaned dataset shape:", df.shape)
print("\nMissing values:")
print(df.isnull().sum())

print("\nReturn/cancellation label counts:")
print(df["returned"].value_counts())

display(df.head())
```

```python

# Overall return/cancellation rate
return_rate = df["returned"].mean() * 100

print(
    f"Return/cancellation transaction-line rate: "
    f"{return_rate:.2f}%"
)

# Product category is not included in this dataset.
# Use country to examine transaction patterns.
country_analysis = df.groupby("Country").agg(
    total_records=("InvoiceNo", "count"),
    return_cancel_rate=("returned", "mean")
)

country_analysis["return_cancel_rate"] *= 100

country_analysis = country_analysis.sort_values(
    "total_records", ascending=False
)

display(country_analysis.head(10))

# Chart 1: Top countries by transaction-line volume
top_countries = country_analysis.head(10)

plt.figure(figsize=(10, 5))
top_countries["total_records"].sort_values().plot(kind="barh")
plt.title("Top 10 Countries by Transaction-Line Volume")
plt.xlabel("Number of records")
plt.ylabel("Country")
plt.tight_layout()
plt.show()

# Chart 2: Return/cancellation rate for the same countries
plt.figure(figsize=(10, 5))
top_countries["return_cancel_rate"].sort_values().plot(kind="barh")
plt.title("Return/Cancellation Rate by Country")
plt.xlabel("Rate (%)")
plt.ylabel("Country")
plt.tight_layout()
plt.show()

# Chart 3: Monthly transaction volume
df["month"] = df["InvoiceDate"].dt.to_period("M").astype(str)

monthly_sales = df.groupby("month").size()

plt.figure(figsize=(12, 5))
monthly_sales.plot(marker="o")
plt.title("Monthly Transaction-Line Volume")
plt.xlabel("Month")
plt.ylabel("Number of records")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

# Build one row per invoice
invoice_df = df.groupby("InvoiceNo").agg(
    invoice_date=("InvoiceDate", "min"),
    country=("Country", "first"),
    line_count=("StockCode", "count"),
    unique_products=("StockCode", "nunique"),
    total_quantity=("Quantity", lambda x: x.abs().sum()),
    average_price=("UnitPrice", "mean"),
    total_value=("total_value", "sum"),
    returned=("returned", "max")
).reset_index()

# Time-based features
invoice_df["month"] = invoice_df["invoice_date"].dt.month
invoice_df["day_of_week"] = invoice_df["invoice_date"].dt.dayofweek
invoice_df["hour"] = invoice_df["invoice_date"].dt.hour

print("Invoice-level dataset:", invoice_df.shape)
print("\nTarget distribution:")
print(invoice_df["returned"].value_counts())

display(invoice_df.head())
```

```python

features = [
    "country",
    "line_count",
    "unique_products",
    "total_quantity",
    "average_price",
    "total_value",
    "month",
    "day_of_week",
    "hour"
]

X = invoice_df[features]
y = invoice_df["returned"]

# Split data by date: older invoices train the model,
# and newer invoices test it.
invoice_df = invoice_df.sort_values("invoice_date")

split_index = int(len(invoice_df) * 0.80)

train_df = invoice_df.iloc[:split_index]
test_df = invoice_df.iloc[split_index:]

X_train = train_df[features]
y_train = train_df["returned"]

X_test = test_df[features]
y_test = test_df["returned"]

numeric_features = [
    "line_count", "unique_products", "total_quantity",
    "average_price", "total_value", "month",
    "day_of_week", "hour"
]

categorical_features = ["country"]

def make_preprocessor():
    return ColumnTransformer([
        ("numeric", Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("scaler", StandardScaler())
        ]), numeric_features),

        ("categorical", Pipeline([
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("encoder", OneHotEncoder(handle_unknown="ignore"))
        ]), categorical_features)
    ])

models = {
    "Logistic Regression": LogisticRegression(
        max_iter=1000,
        class_weight="balanced"
    ),
    "Decision Tree": DecisionTreeClassifier(
        max_depth=6,
        min_samples_leaf=20,
        class_weight="balanced",
        random_state=42
    )
}

trained_models = {}
results = []

for name, model in models.items():

    pipeline = Pipeline([
        ("preprocessing", make_preprocessor()),
        ("model", model)
    ])

    pipeline.fit(X_train, y_train)

    predictions = pipeline.predict(X_test)
    probabilities = pipeline.predict_proba(X_test)[:, 1]

    trained_models[name] = pipeline

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, predictions),
        "Precision": precision_score(
            y_test, predictions, zero_division=0
        ),
        "Recall": recall_score(
            y_test, predictions, zero_division=0
        ),
        "F1 Score": f1_score(
            y_test, predictions, zero_division=0
        ),
        "ROC-AUC": (
            roc_auc_score(y_test, probabilities)
            if y_test.nunique() == 2 else np.nan
        )
    })

    print("\nModel:", name)
    print(classification_report(
        y_test, predictions, zero_division=0
    ))

    ConfusionMatrixDisplay.from_predictions(
        y_test, predictions
    )
    plt.title(f"Confusion Matrix: {name}")
    plt.show()

results_df = pd.DataFrame(results)
display(results_df)
```

```python

tree_model = trained_models["Decision Tree"]

feature_names = (
    tree_model
    .named_steps["preprocessing"]
    .get_feature_names_out()
)

importance = (
    tree_model
    .named_steps["model"]
    .feature_importances_
)

importance_df = pd.DataFrame({
    "Feature": feature_names,
    "Importance": importance
}).sort_values("Importance", ascending=False)

display(importance_df.head(15))

top_features = importance_df.head(15).sort_values(
    "Importance"
)

plt.figure(figsize=(10, 6))
plt.barh(
    top_features["Feature"],
    top_features["Importance"]
)
plt.title("Top 15 Features in Decision Tree")
plt.xlabel("Feature importance")
plt.tight_layout()
plt.show()
```

```python

invoice_df.to_csv("invoice_analysis.csv", index=False)
country_analysis.to_csv("country_analysis.csv")
results_df.to_csv("model_comparison.csv", index=False)
importance_df.to_csv("feature_importance.csv", index=False)

print("All result files created successfully!")
```
